```python
def is_prime(n):
    if n < 2:
        return False

    for divisor in range(2, n):
        if n % divisor == 0:
            return False

    return True
```

```python
assert is_prime(7) is True
assert is_prime(9) is False
assert is_prime(1) is False
assert is_prime(2) is True
print("All 4 prime checks passed.")
```
